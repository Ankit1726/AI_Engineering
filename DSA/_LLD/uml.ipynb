#### ***UML (Unified Modeling Language) Basics***

- Association: Do objects ka aapsi temporary milan.

- Aggregation: Department aur Student (Alag ho sakte hain - Hollow Diamond).

- Composition: Car aur Engine (Sath jiyenge, sath marenge - Filled Diamond).

- Inheritance: Father aur Son / Animal aur Dog (Is-A relationship).

- Dependency: Printer aur Document (Kaam nikla aur bhool gaye - Dotted Arrow).

```python
# INHERITANCE (Is-A Relationship) denoted as (—◁)
class Member:
    def __init__(self,name:str):
        self._name = name 

    def get_name(self) ->str:
        return self._name 

class Student(Member):
    def __init__(self,name:str,roll_no:int):
        super().__init__(name)
        self.roll_no = roll_no
```

```python
# COMPOSITION (Strong Has-A Relationship) denoted as (◆—)
class Brain:
    def think(self):
        return "Thinking deeply about OOP design patterns..."

class Human:
    def __init__(self, name: str):
        self.name = name
        # Composition: Brain ka janam Human ke sath ho raha hai.
        # Agar Human object destroy hua, toh Brain bhi automatically khatam.
        self.brain = Brain()  

    def work(self):
        print(f"{self.name} is {self.brain.think()}")
```

```python
# AGGREGATION (Weak Has-A Relationship) denoted as (◇—)
class Department:
    def __init__(self,dept_name:str):
        self.dept_name = dept_name
        self.students = []
    
    def add_student(self, student: Student):
        self.students.append(student)
    
    def show_students(self):
        print(f"\n--- Students in {self.dept_name} Department ---")
        for s in self.students:
            print(f"- {s.get_name()} (Roll No: {s.roll_no})")
```

```python
# ASSOCIATION (Interaction) denoted as (--->) & DEPENDENCY (Temporary Usage) denoted as (- - ->)

class Document:
    def __init__(self,title:str):
        self.title = title 

class Teacher(Member):
    def __init__(self, name: str):
        super().__init__(name)

    # Association:(Interaction)
    def teach(self, student: Student):
        print(f"\n[Association] Teacher {self.get_name()} is teaching Student {student.get_name()}")
    
    # Dependency
    def prepare_notes(self, doc_title: str) -> Document:
        return Document(doc_title)

    
class Printer:
    def print_document(self, document: Document):
        print(f"[Dependency] Printer is printing notes: '{document.title}'")
```

```python
if __name__ == "__main__":
    print("=== UML Basics Live Execution ===")

    # 1. Inheritance Check
    student1 = Student("Rahul", 101)
    student2 = Student("Priya", 102)
    teacher = Teacher("Sharma Sir")
    print(f"Created Student: {student1.get_name()} using inherited Member properties.")

    # 2. Association Check
    teacher.teach(student1)

    # 3. Aggregation Check
    computer_science = Department("Computer Science")
    computer_science.add_student(student1)
    computer_science.add_student(student2)
    computer_science.show_students()

    # 4. Composition Check
    person = Human("Ankit")
    person.work()

    # 5. Dependency Check
    notes = teacher.prepare_notes("Low Level Design - UML Basics")
    office_printer = Printer()
    office_printer.print_document(notes)
```

```python
# Lifecycle Demonstration (Aggregation vs Composition)
print("\n=== Simulating Object Deletion ===")
del computer_science
print("[Aggregation Proof]: Department deleted, but student objects are still alive!")
print(f"Student 1 is still accessible: {student1.get_name()}")

del person
print("[Composition Proof]: Human object deleted. Internal 'brain' instance closed out from memory completely.")
```
