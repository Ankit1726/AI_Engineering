- ***Encapsulation:Bundling(Attributes+Methods) Single Unit***

```python
class BankAccount:
    def __init__(self,owner:str,bal:int):
        self.owner = owner
        self.__bal = bal  # private var 

    # Getter Method  
    def get_bal(self):
        return self.__bal 
    
    # Setter Method 
    def deposit(self,amount:float):
        if amount > 0:
            self.__bal+=amount
            print(f"successfully deposited ${amount}.")
        else:
            print("invalid deposit amount.")

    def withdraw(self,amount:int):
        if 0<amount<=self.__bal:
            self.__bal -=amount
            print(f"successfully withdraw ${amount}")
        else:
            print(f"insufficient balance.")

acc = BankAccount("Ankit",89898)
print(acc.owner)

acc.deposit(38007)
acc.withdraw(10000)
```

- ***Inheritance: [Child Class --> Parent Class]***

```python
# Parent Class 
class Animal:
    def __init__(self,name:str,age:int):
        self.name = name 
        self.age = age 

    def eat(self):
        print(f"{self.name} is eating.")

    def sleep(self):
        print(f"{self.name} is sleeping")
    
    def move(self):
        print(f"{self.name} is moving around.")

class Dog(Animal):
    def __init__(self, name:str, age:int,breed:str):
        super().__init__(name, age)
        self.breed = breed 

    def bark(self):
        print(f"{self.name} says: Woof..!")
    
    def move(self):
        print(f"{self.name} is running happily on 4 lags")

dog = Dog("Rocky",4,"Germanseffart")

dog.eat()
dog.sleep()
dog.bark()
dog.move()
```

***Polymorphism: (same name but different work)***
- ***single interface behave differently depending on object call it.***

```python
class iPhone:
    def unlock(self):
        print("Scanning Face ID... Unlocked.!")
    
class Android:
    def unlock(self):
        print("Scanning Fingerprint on the back... Unlocked.!")

class OldPhone:
    def unlock(self):
        print("Entering 4-digit PIN... Unlocked!")

def open_open(phone_object):
    phone_object.unlock()

phone1 = iPhone()
phone2 = Android()
phone3 = OldPhone()

open_open(phone1)
open_open(phone2)
open_open(phone3)
```

- ***Abstraction: Hiding internal implementation details showing essential features.***

```python
from abc import ABC,abstractmethod

# Abstract Class (Contract)
class Shape(ABC):
    @abstractmethod
    def area(self):
        """Calculates the area of the shape. 
        Implementation details are hidden here."""
        pass

    @abstractmethod
    def perimeter(self):
        """Calculates the perimeter of the shape."""
        pass

# Actual Implementation
class Rectangle(Shape):
    def __init__(self,length:int,breadth:int):
        self.length = length
        self.breadth = breadth 

    # Overriding abstract method 
    def area(self):
        return self.length * self.breadth
    
    def perimeter(self):
        return 2*(self.length + self.breadth)
    
class Circle(Shape):
    def __init__(self,radius:float):
        self.radius = radius

    def area(self):
        return 3.14 * (self.radius**2)
    
    def perimeter(self):
        return 2*3.14*self.radius
    
shapes = [Rectangle(12,7),Circle(7)]

for s in shapes:
    print(f"Area:{s.area()} & Perimeter:{s.perimeter()}")
```

```python
# Problem Statement

# Create a class Movie with the following:

# Attributes:
# movie_name —> name of the movie
# total_seats —> total seats available in the theatre
# ticket_price —> price per ticket
# booked_seats —> starts at 0

# Methods:
# book_ticket(num_tickets) — books the given number of tickets. If enough seats are available,
# confirm the booking and show the total amount to pay. If not,
# show "Sorry, not enough seats available"

# show_status() — displays movie name, seats available, and seats booked so far
```

```python
class Movie:
    def __init__(self, movie_name: str, total_seats: int, ticket_price: int):
        self.movie_name = movie_name
        self.total_seats = total_seats
        self.ticket_price = ticket_price
        self.booked_seats = 0

    def book_tickets(self,num_tickets:int):
        if num_tickets > self.total_seats - self.booked_seats:
            print('sorry,not seat available..!')
        else:
            self.booked_seats+=num_tickets
            self.total_seats-= num_tickets
            print('your ticket is confirmed..!')
            print(f"Total Price: {self.ticket_price * num_tickets}")
    
    def show_status(self) -> None:
        print(f"Movie Name: {self.movie_name}")
        print(f"Seats Available: {self.total_seats}")
        print(f"Total Bookings: {self.booked_seats}")

movie = Movie("Krish 4",117,477)
movie.show_status()

movie.book_tickets(77)
movie.show_status()
```
