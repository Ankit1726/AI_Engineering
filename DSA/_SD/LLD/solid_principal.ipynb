#### ***Solid Principals***
---
- ***Single Responsibility Principle (SRP)***
   - ***Single job per class.***

```python
class User:
     def __init__(self,name,age,email):
          self.name = name 
          self.age = age 
          self.email = email

     def get_user(self):
          print(f"I am {self.name} & My age is {self.age}")

     def is_adult(self) -> bool:
          return self.age>18
```

```python
class UserRepo:
     def __init__(self,db,user,password):
          self.db = db
          self.user = user
          self.password = password

     def save(self,user:"User"):
          print(f"{user.name} is getting saved to database success..!")

     def delete(self,user:"User"):
          print(f"{user.name} is getting deleted to database success..!")
```

```python
# Main File
user_obj = User("Ankit",22,"anki17@gmail.com")
user_repo = UserRepo("userDB","root","root@123")

user_obj.get_user()
user_repo.save(user_obj)
```

- ***Open/Closed Principle (OCP)***
   - ***Use Abstract classes to add features without modifying working files.***

```python
from abc import ABC,abstractmethod

# Rule or Schema
class  Payment(ABC):
     @abstractmethod
     def pay(self,amount:float):
          pass

class CC_Payment(Payment):
     def pay(self,amount:float):
          print(f"Processing ${amount} via Credit Card...!")

class PayPal(Payment):
     def pay(self,amount:float):
          print(f"Processing ${amount} via PayPal Gateway..!")

class Debit_Card(Payment):
     def pay(self,amount:float):
          print(f"Processing ${amount} via Debit Card..!")

class PaymentProcessor:
     def process_payment(self,payment_method:Payment,amount:float):
          payment_method.pay(amount)
```

```python
payment_process = PaymentProcessor()

upi = PayPal()
cc = CC_Payment()

payment_process.process_payment(cc,4900)
```

- ***Liskov Substitution Principle (LSP)***
   - ***Derived classes must be completely substitutable for their base classes. Yani Child class ka object apne Parent class ke object ko bina kisi crash ya ajeeb behavior ke replace kar sake.***

```python
# Bank Account System 
from abc import ABC,abstractmethod

class Account(ABC):
     def __init__(self,bal):
          self.bal = bal 

     @abstractmethod
     def deposit(self):
          pass
```

```python
from abc import abstractmethod

class Current_Account(Account):
     def __init__(self,bal):
          super().__init__(bal)

     @abstractmethod
     def withdraw(self,amount):
          pass
```

```python
class FD_Account(Account):
     def __init__(self, bal):
          super().__init__(bal)

     def deposit(self,amount):
          self.bal+=amount
          print(f"Amount Deposited Sucessfully..! Your Current Amount:{self.bal}")
          
```

```python
class Saving_Account(Current_Account):
     def __init__(self, amount):
          super().__init__(amount)

     def deposit(self,amount):
          self.bal+=amount
          print(f"Amount Deposited Sucessfully..! Your Current Amount:{self.bal}")

     def withdraw(self,amount):
          if self.bal < amount:
               print("not enough money..!")
          else:
               self.bal-=amount
               print(f"Amount Withdrawl Success..! Your Current Amount:{self.bal}")
```

```python
acc = Saving_Account(789090)
acc.deposit(98789)
acc.withdraw(10000)
```

```python
fd = FD_Account(7000000)
fd.deposit(300000)
```

- ***Interface Segregation Principle (ISP)***
   - ***Break huge interfaces into micro-interfaces.***

```python
from abc import ABC,abstractmethod

class Workable(ABC):
     @abstractmethod
     def work(self):
          print

class Eatable(ABC):
     @abstractmethod
     def eat(self):
          pass 

class Robot(Workable):
     def work(self):
          print("Robot is working..!")

class Employee(Workable,Eatable):
     def eat(self):
          print("employee is eating..")

     def work(self):
          print("employee is working..")

e = Employee()
e.eat()
e.work()

r = Robot()
r.work()
```

- ***Dependency Inversion Principle (DIP)***
   - ***High-level modules should not depend on low-level modules. Both should depend on abstractions." Core logic ko direct low-level classes (Database/APIs) se mat jodo, beech me ek bridge (interface) rakho.***

```python
from  abc import  ABC,abstractmethod

class Notification_Channel(ABC):
     @abstractmethod
     def send(self,msg):
          pass

class Notification_Service:
     def __init__(self,channel:Notification_Channel):
          self.channel = channel

     def notify(self,msg):
          self.channel.send(msg)

class SMS_Service(Notification_Channel):
     def send(self,msg):
          print(f"sending sms:{msg}")

class Email_Service(Notification_Channel):
     def send(self,msg):
          print(f"sending email:{msg}")
```

```python
sms = SMS_Service()
ns = Notification_Service(sms)
ns.notify("Hello Engineers")
```

```python
email = Email_Service()
e = Notification_Service(email)
e.notify("Hello AIML Engineer")
```
