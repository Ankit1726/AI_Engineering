##### ***Singleton Design Pattern***
   - ***The Singleton Design Pattern is a creational pattern that ensures a class has only one instance throughout the entire application lifecycle and provides a global point of access to it.***

```python
class Logger:
     _instance = None
     
     def __new__(cls,file_name:str):
          if cls._instance == None:
               cls._instance = super().__new__(cls)
               cls._instance.file_name = file_name 
               cls._instance.log_count = 0
               return cls._instance
          else:
               return cls._instance

     def log(self,message):
          print(f"Logging {message} in {self.file_name}")
          self.log_count+=1

     def get_count(self)-> int:
          return self.log_count

     
log1 = Logger("app.log")
log2 = Logger("app.log")
log3 = Logger("app.log")
print(f"Are instances identical? {log1 is log2 is log3}")

log1.log("User logged in.")
log2.log("Transaction completed.")
log3.log("Server error occurred.")

print(f"Total global logs recorded: {log1.get_count()}")
```

```python
log1 = Logger("app.log")
log1.log("Hey")

log2 = Logger("app.log")
log2.log("Bye")

log3 = Logger("app.log")
log3.log("Good")

print(log1.get_count())
print(log2.get_count())
print(log3.get_count())
```
