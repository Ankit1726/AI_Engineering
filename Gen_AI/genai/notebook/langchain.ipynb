### ***Mini ChatBot Example***

```python
from langchain_groq import ChatGroq
from langchain_core.messages import AIMessage,SystemMessage,HumanMessage

llm = ChatGroq(
     model = "openai/gpt-oss-120b",
     temperature=0.9,
     streaming=True
)

choice = int(input("tell your response :"))
if choice == 1:
    mode = "You are an angry AI agent. You respond aggressively and impatiently."
elif choice == 2:
    mode = "You are a very funny AI agent. You respond with humor and jokes."
elif choice == 3:
    mode = "You are a very sad AI agent. You respond in a depressed and emotional tone."


messages = [
    SystemMessage(content=mode)
]
print("----------------- welcom type 0 to exit the application-----------------")
while True:
    
    prompt = input("You : ")
    messages.append(HumanMessage(content=prompt))
    if prompt == "0":
        break
    response = llm.invoke(messages)
    messages.append(AIMessage(content=response.content))
    print("Bot :",response.content)
print(messages)
```

### ***Vector Database***
- ***A Vector Database is a specialized storage engine designed to store, index, and query high-dimensional numerical representations (embeddings) of unstructured data like text, images, or audio, enabling fast semantic similarity search rather than exact keyword matching.***

```python
from langchain_community.vectorstores import Chroma 
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_core.documents import Document
from dotenv import load_dotenv 
load_dotenv()
```

```python
docs = [
    Document(page_content="What is RAG and its Components.", metadata={"source": "blog.txt"}),
    Document(page_content="what is transformer.", metadata={"source": "transformer.pdf"})
]
```

```python
# create embedding
embedding_model = HuggingFaceEmbeddings(
    model_name="all-MiniLM-L6-v2"
)

# vector database creation 
vectorstore = Chroma.from_documents(
     documents=docs, 
     embedding=embedding_model,
     persist_directory="../data/chroma-db"
)
```

```python
# pperform simialrity search 
retriever = vectorstore.as_retriever(search_kwargs={"k":2})
retrieved_docs = retriever.invoke("Explain RAG Systems")

for d in retrieved_docs:
    print(f"Content: {d.page_content}")
    print(f"Metadata: {d.metadata}")
```

### ***Retriever***
- ***Retriever is a component that accepts an un-indexed query (like a user question) and searches through a data source such as a vector database, search engine, or API to return the most relevant documents or context snippets.***

***1.Arxiv Retriever***

```python
import arxiv
from langchain_core.documents import Document

def get_arxiv_docs(query: str, max_docs: int = 3):
    client = arxiv.Client()
    search = arxiv.Search(
        query=query,
        max_results=max_docs,
        sort_by=arxiv.SortCriterion.Relevance
    )
    
    documents = []
    # Latest arxiv syntax (v2.0+)
    for result in client.results(search):
        doc = Document(
            page_content=result.summary,
            metadata={
                "Title": result.title,
                "Authors": [author.name for author in result.authors],
                "Published": str(result.published),
                "Entry_ID": result.entry_id
            }
        )
        documents.append(doc)
    return documents

# 1. Fetch papers
docs = get_arxiv_docs("large Language Models (LLM)", max_docs=3)

# 2. Print results
for i, doc in enumerate(docs):
    print(f"\nResult {i+1}")
    print("Title:", doc.metadata.get("Title"))
    print("Authors:", doc.metadata.get("Authors"))
    print("Summary:", doc.page_content)
```

#### ***Runnable***
- ***A Runnable is a fundamental protocol/interface in LangChain (LCEL) that standardizes execution methods like invoke, stream, and batch across all components (prompts, models, parsers, chains).***

#### ***RunnableSequence***
- ***It is a component in LCEL (LangChain Expression Language) that chains multiple Runnables sequentially, where the output of each step is automatically passed as the input to the next step.***

```python
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# Prompt Template 
prompt = ChatPromptTemplate.from_template(
    "Explain {topic} in simple words"
)

# LLM
model = ChatGroq(model="openai/gpt-oss-20b")

# output parser
parser = StrOutputParser()

# chains 
chain = prompt | model | parser 
result = chain.invoke("What is deep neural networks")
print(result)
```

#### ***Runnable Passthrough***
- ***It is a special Runnable component that passes its input data unchanged (as-is) to the next step or key in a chain.***

- ***RunnablePassthrough.assign() :It dynamically add new keys to an existing input dictionary without overwriting the original input fields***

```python
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableParallel,RunnablePassthrough

model = ChatGroq(model="openai/gpt-oss-20b")
parser = StrOutputParser()

code_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a code generator"),
    ("human", "{topic}")
])

explain_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant who explains code in simple terms"),
    ("human", "Explain the following code in simple words:\n{code}")
])

sequence = code_prompt | model | parser
```

```python
seq = RunnableParallel(
     {
     "code":RunnablePassthrough(),
     "explanation" : explain_prompt | model | parser
     }
)
chain = sequence | seq

result = chain.invoke({"topic" : "please write a code of fibonacci code in python "})
print(result['code'])
print(result['explanation'])
```

```python
from langchain_core.runnables import RunnableParallel,RunnableLambda
# Two different prompts
short_prompt = ChatPromptTemplate.from_template(
    "Explain {topic} in 1-2 lines"
)

detailed_prompt = ChatPromptTemplate.from_template(
    "Explain {topic} in detail"
)

topic = "Deep Learning Neural Networks"

chain = RunnableParallel({
    "short" :RunnableLambda(lambda x :x['short']) |short_prompt | model | parser ,
    "detailed" :RunnableLambda(lambda x: x['detailed']) |detailed_prompt |model |parser
})
```

```python
result = chain.invoke({
    "short" : {"topic":"Transformere"},
    "detailed" : {"topic":"LLM's"}
})

print(result['short'])
print(result['detailed'])
```

### ***Tool Calling***
- ***It is a capability in LLMs where the model detects when external data or actions are needed and outputs structured JSON containing function names and arguments instead of a plain text response.***

```python
from langchain_groq import ChatGroq
from langchain.tools import tool
from langchain_core.messages import HumanMessage,ToolMessage
from rich import print 

# create tool 
@tool
def greet_tool(name:str) -> str:
     """return a creative greet message with name"""
     return f"Hello {name} Welcome to Indus emperor"

tools = {
     "greet_tool":greet_tool
}

llm = ChatGroq(model="openai/gpt-oss-20b")

# tool binding 
llm_tool = llm.bind_tools([greet_tool])
```

```python
message = []
prompt = input("You: ")
query = HumanMessage(prompt)
message.append(query)

# answer 
result = llm_tool.invoke(message)
message.append(result)

if result.tool_calls:
    for tool_call in result.tool_calls:
        tool_name = tool_call["name"]
        
        # Tool ko execute karo
        tool_output = tools[tool_name].invoke(tool_call["args"])
        
        # Result ko ToolMessage format mein convert karke array mein add karo
        tool_message = ToolMessage(
            content=str(tool_output),
            tool_call_id=tool_call["id"]
        )
        message.append(tool_message)
   
    # Second LLM call - synthesizes final answer using tool response
    result = llm_tool.invoke(message)
print(result.content)
```

```python
from langchain_tavily import TavilySearch
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

search_tool = TavilySearch(max_result = 5)
llm = ChatGroq(
     model = "openai/gpt-oss-20b"
)

prompt = ChatPromptTemplate.from_template(
     """
     You are a helpful assistant
     summarize the following news into clear bullet points
     {news}
"""
)
chain = prompt | llm | StrOutputParser()
```
