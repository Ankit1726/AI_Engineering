### ***Behavioral Design Pattern***
---
- ***Memento Design Pattern***
    - ***Memento ek Behavioral Design Pattern hai jiska use bina object ki encapsulation (privacy) ko tode, uski internal state ka ek snapshot capture karne ke liye kiya jata hai, taaki use baad me Undo/Redo ya Rollback kiya ja sake.***
    - ***Main advantage it follows Single Responsibility Principle (SRP).***

```python
class TextMemento:
     def __init__(self,text):
          self.__saved_text = text 

     def get_saved_text(self):
          return self.__saved_text
```

```python
class TextEditor:
     def __init__(self):
          self.__text = ""

     def write(self,new_txt):
          self.__text+=new_txt

     def get_txt(self) -> str:
          return self.__text
     
     def save(self) -> TextMemento:
          return TextMemento(self.__text)

     def restore(self,tm:TextMemento):
          self.__text = tm.get_saved_text()
```

```python
from typing import List 
class History:
     def __init__(self):
          self.__history:List[TextMemento] = []
     
     def save_state(self,tm:TextMemento):
          self.__history.append(tm)

     def undo(self) -> TextMemento:
          if len(self.__history) > 0:
            self.__history.pop()
            if len(self.__history) == 0:
                return TextMemento("")
            return self.__history[-1]
          else:
               return TextMemento("")

     def get_history(self):
        for i in range(len(self.__history)):
            print(f"{i} = {self.__history[i].get_saved_text()}")
```

```python
text_editor = TextEditor()
history = History()

text_editor.write("Hello")
text_editor.write(" World")
history.save_state(text_editor.save())
print(text_editor.get_txt())
```

```python
text_editor.write("Good")
text_editor.write(" Morning")
history.save_state(text_editor.save())
print(text_editor.get_txt())
```

```python
text_editor.restore(history.undo())
print(text_editor.get_txt())
```

- ***Oberver Pattern***
  - ***Observer ek Behavioral Design Pattern hai jo objects ke beech ek One-to-Many dependency create karta hai. Jab ek object (Subject) ki state change hoti hai, toh usse dependent saare objects (Observers) ko automatically notify aur update kar diya jata hai.***

```python
from abc import ABC,abstractmethod

# Observer Iterface 
class Observer(ABC):
     @abstractmethod
     def update(self,temperature):
          pass 

# Concrete Observer
class TV(Observer):
     def update(self, temperature):
          print(f"TV Display: Temperature updated to {temperature}°C")

class Mobile(Observer):
     def update(self, temperature):
          print(f"Mobile Display: Temperature updated to {temperature}°C")
```

```python
# Wheather Station
class WheatherStation:
     def __init__(self):
          self._observers = []
          self._temperature = 0

     def add_observer(self, observer: Observer):
        self._observers.append(observer)

     def remove_observer(self, observer: Observer):
        self._observers.remove(observer)

     def set_temperature(self, new_temp):
        self._temperature = new_temp
        self.notify_observers()  # Automatically notify everyone

     def notify_observers(self):
        for observer in self._observers:
            observer.update(self._temperature)

station = WheatherStation()
tv = TV()
mobile = Mobile()
```

```python
station.add_observer(tv)
station.add_observer(mobile)


print("--- First Update ---")
station.set_temperature(30) 
station.remove_observer(tv)

print("\n--- Second Update (TV Removed) ---")
station.set_temperature(35)
```

- ***Strategy Design Pattern [Provide multiple algorithm ao choose a best Of one]***

  - ***Iye ek Behavioral Pattern hai jo interchangeable algorithms (ya business logics) ki ek family define karta hai, aur unhe runtime par dynamic tarike se switch karne ki flexibility deta hai.***

  - ***It follows Open-Closed Principal (OCP)***

```python
from abc import ABC,abstractmethod

# Strategy Interface 
class DiscountStrategy(ABC):
     @abstractmethod
     def calculate_discount(self):
          pass

# Concrete Strategies
class Diwali(DiscountStrategy):
     def calculate_discount(self):
          print("Applying 20% Diwali Discount")

class Holi(DiscountStrategy):
     def calculate_discount(self):
          print("Applying 10% Holi Discount")
```

```python
# Context 
class DiscountService:
     def __init__(self,strategy:DiscountStrategy):
          self.strategy = strategy

     def set_strategy(self,strategy:DiscountStrategy):
          self.strategy = strategy

     def process(self):
          self.strategy.calculate_discount()

service = DiscountService(Diwali())
service.process()

service.set_strategy(Holi())
service.process()
```

- ***Command Design Pattern***
   - ***Iye ek Behavioral Pattern hai jo ek request ya operation ko ek standalone object (Command) me encapsulate kar deta hai, jisse client ko bina exact receiver aur parameters jane operations ko execute, queue, ya undo karne ki flexibility milti hai.***

```python
from abc import ABC,abstractmethod

# Reciever
class Chef:
     def cook_pizza(self):
          print("Chef is cooking Pizza!")
     def cook_burger(self):
          print("Chef is cooking Burger!")

# Command Interface 
class Order(ABC):
     @abstractmethod
     def execute(self):
          pass

# Concrete Command 
class PizzaOrder(Order):
     def __init__(self,chef:Chef):
          self.chef = chef

     def execute(self):
          self.chef.cook_pizza()
```

```python
class BurgerOrder(Order):
     def __init__(self,chef:Chef):
          self.chef = chef 

     def execute(self):
          self.chef.cook_burger()

# Invoker 
class Waiter:
     def take_order(self,order:Order):
          order.execute()

chef = Chef()
waiter = Waiter()


# Placing Pizza Order
pizza_order = PizzaOrder(chef)
waiter.take_order(pizza_order)

# Placing Burger Order
burger_order = BurgerOrder(chef)
waiter.take_order(burger_order)
```

- ***Template Pattern***
   - ***Iye ek Behavioral Pattern hai jo ek algorithm ke skeleton (structure) ko base class ya abstract class me define karta hai, aur uske specific steps ko subclasses me bina algorithm ka structure badle override karne ki flexibility deta hai.***

```python
from abc import ABC,abstractmethod

# Abstract Class 
class DataParser(ABC):
     def open_file(self):
          print("Opening the file...")

     def close_file(self):
          print("Closing the file...")

     @abstractmethod
     def parse_data(self):
          pass 

     # Template Method 
     def parse(self):
          self.open_file()
          self.parse_data()
          self.close_file()
```

```python
# Concrete Subclass 
class CSVParser(DataParser):
     def parse_data(self):
          print("Parsing CSV file data specific logic.")

class JSONParser(DataParser):
     def parse_data(self):
          print("Parsing JSON file data specific logic.")


print("--- CSV Processing ---")
csv_processor = CSVParser()
csv_processor.parse()

print("\n--- JSON Processing ---")
json_processor = JSONParser()
json_processor.parse()
```

#### ***Iterator Pattern***
   - ***It provides a standard way to sequentially access elements of a collection sequentially without exposing its underlying internal structure or implementation details (like arrays, lists, or sets).***

```python
from abc import ABC,abstractmethod

class Song:
    def __init__(self,title:str):
        self._title = title

    def get_title(self) -> str:
        return self._title
    
class Iterator(ABC):
    @abstractmethod
    def has_next(self)-> bool:
        pass 
    def next(srlf)-> Song:
        pass  
```

```python
class Playlist_Itr(Iterator):
    def __init__(self,songs:list):
        self._songs = songs 
        self._position = 0

    def has_next(self) -> bool:
        return self._position < len(self._songs)
    
    def next(self) -> Song:
        if self.has_next():
            song = self._songs[self._position]
            self._position += 1
            return song
        return None
```

```python
class Playlist:
    def __init__(self):
        self._songs = []

    def add_song(self,song:Song):
        self._songs.append(song)

    def create_iterator(self) -> Playlist_Itr:
        return Playlist_Itr(self._songs)
    
playlist = Playlist()
playlist.add_song(Song("Song A"))
playlist.add_song(Song("Song B"))
playlist.add_song(Song("Song C"))

iterator = playlist.create_iterator()
while iterator.has_next():
    song = iterator.next()
    print(song.get_title())
```

#### ***State Design Pattern*** 
   - ***The State Design Pattern is a behavioral pattern that allows an object to alter its behavior when its internal state changes, making the object appear as if it changed its entire class.***

```python
from abc import ABC,abstractmethod

# State Interface 
class TransportMode(ABC):
    @abstractmethod
    def compute_eta(self) -> str:
        pass 

    @abstractmethod
    def get_direction(self) -> str:
        pass 
```

```python
class Bike_Mode(TransportMode):
    def compute_eta(self) -> str:
        return "Bike will take 15 minutes."
    
    def get_direction(self)-> str:
        return "Go left to the road."
    
class Walk_Mode(TransportMode):
    def compute_eta(self) -> str:
        return "Walk will take 30 minutes."
    
    def get_direction(self) -> str:
        return "walk right to the road then take left"
```

```python
class Direction_Service:
    def __init__(self,initial_mode:TransportMode):
        self._current_mode = initial_mode

    def change_mode(self,new_mode:TransportMode):
        self._current_mode = new_mode

    def show_eta(self):
        print(self._current_mode.compute_eta())

    def show_direction(self):
        print(self._current_mode.get_direction())

service = Direction_Service(Bike_Mode())
service.show_eta()
service.show_direction()

service.change_mode(Walk_Mode())
service.show_eta()
service.show_direction()
```

#### ***Mediator Design Pattern***
  -  ***The Mediator Design Pattern is a behavioral pattern that defines an object (the mediator) to encapsulate and control how a set of objects interact, preventing them from referring to each other directly and promoting loose coupling.***

```python
from abc import ABC, abstractmethod 

# 1. Mediator Interface 
class ChatMediator(ABC):  # Fixed typo: ChaMediator -> ChatMediator
     @abstractmethod
     def send_message(self, message: str, user: "User"):
          pass 

# 2. Concrete Mediator 
class ChatRoom(ChatMediator):
     def __init__(self):
          self._users = []

     def add_user(self, user: "User"):
          self._users.append(user)
          user.set_mediator(self)

     def send_message(self, message: str, sender: "User"):
          for user in self._users:
               if user != sender:
                    user.receive(message, sender.name)
```

```python
# 3. Colleague Class
class User:
     def __init__(self, name: str):
          self.name = name
          self.mediator = None 

     def set_mediator(self, mediator: ChatMediator):
          self.mediator = mediator  # Fixed typo: seld -> self

     def send(self, message: str):
          print(f"--- {self.name} sends: '{message}' ---")
          if self.mediator:
               self.mediator.send_message(message, self)

     def receive(self, message: str, sender_name: str):
          print(f"[{self.name}'s Screen] Received from {sender_name}: '{message}'")


# --- Testing Execution ---
chat = ChatRoom()

kavi = User("Kavi")
ankit = User("Ankit")

# Register Users
chat.add_user(kavi)
chat.add_user(ankit)

kavi.send("Hello Ankit, Meet At 7PM")
ankit.send("Ok Done..!")
```
