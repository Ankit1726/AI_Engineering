#### ***Synchromous Programming***
- Tasks are executed one after the other. If Task 1 is waiting for an API response, the entire program halts (blocks) and waits until it finishes before starting Task 2.

```python
import time 

def fetch_wheather():
    print("fetching wheather data..!")
    time.sleep(3)
    print("data feteched sucess..!")

def fetch_news():
    print("fetching news data..!")
    time.sleep(4)
    print("data feteched sucess")

def main():
    st_time = time.time()
    fetch_wheather()
    fetch_news()

    ed_time = time.time()
    print(f"total time taken:{ed_time-st_time} sec")
```

```python
main()
```

#### ***Asynchronous Programming***
- Allows the program to move on to another task while waiting for a slow external operation (like an LLM API or web search) to complete.

```python
import time 
import asyncio

async def fetch_wheather():
    print("fetching wheather data..!")
    await asyncio.sleep(2) # simulate network delay
    print("data feteched sucess..!")

async def fetch_news():
    print("fetching news data..!")
    await asyncio.sleep(3)
    print("data feteched sucess")

async def main():
    st_time = time.time()
    await asyncio.gather(fetch_wheather(),fetch_news())
    ed_time = time.time()

    print(f"total time taken:{ed_time-st_time} sec")
```

```python
await main()
```
